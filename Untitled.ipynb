```python
import numpy as np
import matplotlib.pyplot as plt
import sys

from mpl_toolkits.mplot3d import Axes3D
from pathlib import Path

ROOT_DIR = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(ROOT_DIR) not in sys.path:
    sys.path.insert(0, str(ROOT_DIR))
    
from trissolaris.config_loader import load_config
from trissolaris.tpo_system_physics import normalize_center_of_mass
from trissolaris.tpo_system_velocity_verlet import VelocityVerletIntegrator
from trissolaris.tpo_system_lyapunov import compute_lyapunov_exponents


plt.rcParams.update({'font.size': 11, 'figure.titlesize': 14})
```

```python
cfg = load_config(ROOT_DIR / "config" / "config_sampling.yaml")
planets = cfg["planets_parsed"]

print(f"Total de planetas carregados: {planets['count']}")
print(f"Shape das posições: {planets['positions'].shape}")
print(f"Shape das velocidades: {planets['velocities'].shape}")

q_p = planets["positions"]
v_p = planets["velocities"]

theta = np.arctan2(q_p[:, 1], q_p[:, 0])
phi = np.arctan2(v_p[:, 1], v_p[:, 0])

print("\nExemplo dos 3 primeiros planetas:")
for i in range(min(3, planets["count"])):
    print(
        f"Planeta {i}: r = {np.linalg.norm(q_p[i]):.3f}, θ = {theta[i]:.3f} rad | "
        f"v = {np.linalg.norm(v_p[i]):.3f}, φ = {phi[i]:.3f} rad"
    )
```

```python
sim_cfg = cfg["simulation"]
dt = sim_cfg["dt"]
T_max = sim_cfg["T_max"]
eps = sim_cfg["eps"]
R_corte = sim_cfg["R_corte"]

m_s = cfg["stars"]["masses"]
R_s = cfg["stars"]["radii"]
q_s = cfg["stars"]["positions"]
v_s = cfg["stars"]["velocities"]

names = planets["names"]
K = planets["count"]

q_s_cm, v_s_cm = normalize_center_of_mass(m_s, q_s, v_s)

R_cm = q_s[0] - q_s_cm[0]
V_cm = v_s[0] - v_s_cm[0]

q_p_cm = q_p - R_cm
v_p_cm = v_p - V_cm
```

```python
integrator = VelocityVerletIntegrator(m_stars=m_s, eps=eps)

results = integrator.simulate_with_events(
    q_s0=q_s_cm, v_s0=v_s_cm, q_p0=q_p_cm, v_p0=v_p_cm, T_max=T_max, R_stars=R_s
)
```

```python
q_s_hist = results["q_s_hist"]
q_p_hist = results["q_p_hist"]
t_series = results["time_array"]
T_sob = results["t_sob"]
events = results["events"]
max_err = results["max_energy_error"]

print(f"=== RELATÓRIO DA REALIZAÇÃO (T_max = {T_max}) ===")
print(f"{'Planeta':<12} | {'Tempo Sobrev. (T_sob)':<22} | {'Status Final':<20}")
print("-" * 60)

for i in range(K):
    status = events[i] if isinstance(events, (list, np.ndarray)) else "Finalizado"
    print(f"{names[i]:<12} | {T_sob[i]:<22.3f} | {status}")
```

```python
fig, ax = plt.subplots(figsize=(9, 8))

# Plot das estrelas
star_colors = ['#d62728', '#ff7f0e', '#bcbd22']
for i in range(len(m_s)):
    ax.plot(q_s_hist[:, i, 0], q_s_hist[:, i, 1], label=f'Estrela {i} (m={m_s[i]})', color=star_colors[i], lw=1.5)
    ax.scatter(q_s_hist[-1, i, 0], q_s_hist[-1, i, 1], color=star_colors[i], s=100, zorder=5)

# Plot dos planetas
names = cfg["planets_parsed"]["names"]
for k in range(q_p_hist.shape[1]):
    # Ignora os valores NaN para desenhar apenas até ao momento do evento
    valid_mask = ~np.isnan(q_p_hist[:, k, 0])
    ax.plot(q_p_hist[valid_mask, k, 0], q_p_hist[valid_mask, k, 1], 
            label=f'{names[k]} ({results["events"][k]})', linestyle='--', alpha=0.8)

ax.set_title("Evolução Temporal do Sistema Trissolaris (Plano XY)", pad=12)
ax.set_xlabel("x [U.A.]")
ax.set_ylabel("y [U.A.]")
ax.axis("equal")
ax.legend(loc="upper right", frameon=True)
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# --- GRÁFICO 1: Trajetórias no Plano XY ---
ax1 = axes[0]

# Trajetória das Estrelas
star_colors = ['#d62728', '#ff7f0e', '#8c564b']
for s in range(q_s_hist.shape[1]):
    ax1.plot(q_s_hist[:, s, 0], q_s_hist[:, s, 1], color=star_colors[s], alpha=0.6, lw=1.2, label=f'Estrela {s+1}')
    ax1.scatter(q_s_hist[0, s, 0], q_s_hist[0, s, 1], color=star_colors[s], s=60, zorder=5)

# Trajetória dos Planetas (filtrando valores NaN para os ejetados)
for p in range(q_p_hist.shape[1]):
    mask = ~np.isnan(q_p_hist[:, p, 0])
    x_p = q_p_hist[mask, p, 0]
    y_p = q_p_hist[mask, p, 1]
    
    if events[p] == "Sobreviveu":
        ax1.plot(x_p, y_p, lw=1.5, label=f'p_{p} (Sobreviveu)', zorder=4)
    else:
        ax1.plot(x_p, y_p, ':', alpha=0.35, lw=0.8)

ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.set_title(f'Trajetórias (Erro máx. energia: {max_err:.2e})')
ax1.set_aspect('equal', 'box')
ax1.grid(True, linestyle=':', alpha=0.5)
ax1.legend(loc='upper right', fontsize=8, framealpha=0.8)

# --- GRÁFICO 2: Tempo de Sobrevivência ---
ax2 = axes[1]
bar_colors = ['#2ca02c' if ev == "Sobreviveu" else '#d62728' for ev in events]
bars = ax2.bar(range(len(T_sob)), T_sob, color=bar_colors, alpha=0.85, edgecolor='black', linewidth=0.5)

ax2.axhline(sim_cfg["T_max"], color='black', linestyle='--', label=f'T_max ({sim_cfg["T_max"]})')
ax2.set_xticks(range(len(T_sob)))
ax2.set_xticklabels([f"p_{i}" for i in range(len(T_sob))], rotation=45)
ax2.set_ylabel('Tempo de Sobrevivência $T_{sob}$')
ax2.set_title('Status Final de Sobrevivência')
ax2.grid(True, axis='y', linestyle=':', alpha=0.6)
ax2.legend()

plt.tight_layout()
plt.show()
```
